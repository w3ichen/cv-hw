## Q1.2 Homography Theory Questions

### Q1.2.1 (5 points)

Prove that there exists a homography **H** that satisfies $\textbf{x}_1 ≡ \textbf{Hx}_2$, given two cameras separated by a pure rotation.

---

YOUR ANSWER HERE...

We're given the camera projection equations:
$$x_1 = K_1 [I  0] X$$
$$x_2 = K_2 [R  0] X$$

1. From $x_2 = K_2R\tilde{X}$ (where $\tilde{X}$ is the 3D point), we get $\tilde{X} \propto R^{-1}K_2^{-1}x_2$
2. Sub into $x_1 = K_1\tilde{X}$: $x_1 \propto K_1R^{-1}K_2^{-1}x_2$
3. Therefore, $x_1 \equiv Hx_2$, where $H = K_1R^{-1}K_2^{-1}$
---

### Q1.2.2 (5 points):

Show that $\textbf{H}^2$ is the homography corresponding to a rotation of 2$\theta$.

---

YOUR ANSWER HERE...

Applying $2\theta$ is just applying the same rotation $\theta$ twice, so it would look like multiplying by $H$ twice: 
1. Rotating first by $\theta$: $x1 = Hx_2$
2. Rotating again by $\theta$ to get to $2\theta$: $x1 = H*(Hx_2)$
3. This can be simplified into $H^2$: $x_1 = H^2x_2$

```python
import os
import numpy as np
import cv2
import skimage.color
import pickle
from matplotlib import pyplot as plt
import scipy
from skimage.util import montage
import time

PATCHWIDTH = 9

def read_pickle(path):
    with open(path, "rb") as f:
        return pickle.load(f)

def write_pickle(path, data):
    with open(path, "wb") as f:
        pickle.dump(data, f)

def briefMatch(desc1,desc2,ratio):

    matches = skimage.feature.match_descriptors(desc1,desc2,
                                                'hamming',
                                                cross_check=True,
                                                max_ratio=ratio)
    return matches

def plotMatches(img1,img2,matches,locs1,locs2):

    fig, ax = plt.subplots(nrows=1, ncols=1)
    img1 = cv2.cvtColor(img1, cv2.COLOR_BGR2GRAY)
    img2 = cv2.cvtColor(img2, cv2.COLOR_BGR2GRAY)
    plt.axis('off')
    skimage.feature.plot_matches(ax,img1,img2,locs1,locs2,
                                 matches,matches_color='r',only_matches=True)
    plt.show()
    return

def makeTestPattern(patchWidth, nbits):

    np.random.seed(0)
    compareX = patchWidth*patchWidth * np.random.random((nbits,1))
    compareX = np.floor(compareX).astype(int)
    np.random.seed(1)
    compareY = patchWidth*patchWidth * np.random.random((nbits,1))
    compareY = np.floor(compareY).astype(int)

    return (compareX, compareY)

def computePixel(img, idx1, idx2, width, center):

    halfWidth = width // 2
    col1 = idx1 % width - halfWidth
    row1 = idx1 // width - halfWidth
    col2 = idx2 % width - halfWidth
    row2 = idx2 // width - halfWidth
    return 1 if img[int(center[0]+row1)][int(center[1]+col1)] < img[int(center[0]+row2)][int(center[1]+col2)] else 0

def computeBrief(img, locs):

    patchWidth = 9
    nbits = 256
    compareX, compareY = makeTestPattern(patchWidth,nbits)
    m, n = img.shape

    halfWidth = patchWidth//2

    locs = np.array(list(filter(lambda x: halfWidth <= x[0] < m-halfWidth and halfWidth <= x[1] < n-halfWidth, locs)))
    desc = np.array([list(map(lambda x: computePixel(img, x[0], x[1], patchWidth, c), zip(compareX, compareY))) for c in locs])

    return desc, locs

def corner_detection(img, sigma):

    # fast method
    result_img = skimage.feature.corner_fast(img, n=PATCHWIDTH, threshold=sigma)
    locs = skimage.feature.corner_peaks(result_img, min_distance=1)
    return locs

def loadVid(path):

    # Create a VideoCapture object and read from input file
    # If the input is the camera, pass 0 instead of the video file name

    cap = cv2.VideoCapture(path)

    # get fps, width, and height
    fps = cap.get(cv2.CAP_PROP_FPS)
    width  = cap.get(cv2.CAP_PROP_FRAME_WIDTH)
    height = cap.get(cv2.CAP_PROP_FRAME_HEIGHT)

    # Append frames to list
    frames = []

    # Check if camera opened successfully
    if cap.isOpened()== False:
        print("Error opening video stream or file")

    # Read until video is completed
    while(cap.isOpened()):

        # Capture frame-by-frame
        ret, frame = cap.read()

        if ret:
            #Store the resulting frame
            frames.append(frame)
        else:
            break

    # When everything done, release the video capture object
    cap.release()
    frames = np.stack(frames)

    return frames, fps, width, height
```

### Q2.1.1 (5 points):

How is the FAST detector different from the Harris corner detector that you've seen in the lectures? Can you comment on its computation performance compared to the Harris corner detector?

---

YOUR ANSWER HERE...

FAST = Features from Accelerated Segment Test

FAST differs from Harris in its detection method and computational approach. While Harris computes horizontal and vertical gradients and filters them with a Gaussian filter to calculate corner strength, FAST uses a simple circle of 16 pixels around a candidate point and detects corners by comparing pixel intensities.

FAST is significantly faster than Harris because it uses simple intensity comparisons rather than complex gradient computations and Gaussian filtering, making it highly suitable for real-time video processing applications.

On average, only 3.8 pixels need to be checked per candidate corner using FAST's high-speed test, compared to more extensive computations required by Harris.

---

### Q2.1.2 (5 points):

How is the BRIEF descriptor different from the filterbanks you've seen in the lectures? Could you use any one of the those filter banks as a descriptor?

---

YOUR ANSWER HERE...

BRIEF = Binary Robust Independent Elementary Features

BRIEF creates binary descriptors by selecting pairs of pixel locations in a smoothed image patch and comparing their intensities to produce a binary string, where each bit represents whether one pixel is brighter than another.

In contrast, filter banks are collections of filters (like Gabor filters) convolved with the image to extract responses at different frequencies and orientations.

Filter bank responses can be used as descriptors, since local descriptors extracted uniformly from images can be seen as banks of non-linear filters.
However, traditional filter bank descriptors like SIFT use 128-dimensional floating-point vectors requiring 512 bytes, while BRIEF uses compact binary strings that are much more memory-efficient and faster to match.
Filter banks could work as descriptors but would be less compact and computationally efficient than BRIEF.

### Q2.1.3 (5 points):

Describe how the Hamming distance and Nearest Neighbor can be used to match interest points with BRIEF descriptors. What benefits does the Hamming distance have over a more conventional Euclidean distance measure in our setting?

---

YOUR ANSWER HERE...

Hamming distance measures the number of positions at which corresponding symbols differ between two binary strings of equal length. For binary descriptors like BRIEF, similarity is measured using Hamming distance, which can be computed very efficiently with a bitwise XOR operation followed by a bit count.

Nearest neighbor matching takes each descriptor from one image and compares it to all descriptors in the second image using a distance metric, then selects the descriptor with the smallest distance as the match. This brute-force approach compares every feature in the first image with all features in the second image to find the best matches.

Benefits of Hamming Distance over Euclidean Distance: Hamming distance is very fast because modern CPUs provide specific instructions (like XOR and bit count operations in SSE instruction sets) to perform these calculations extremely efficiently.

```python
def matchPics(I1, I2, ratio, sigma):
    """
    Match features across images

    Input
    -----
    I1, I2: Source images (RGB or Grayscale uint8)
    ratio: ratio for BRIEF feature descriptor
    sigma: threshold for corner detection using FAST feature detector

    Returns
    -------
    matches: List of indices of matched features across I1, I2 [p x 2]
    locs1, locs2: Pixel coordinates of matches [N x 2]
    """

    # ===== your code here! =====

    # TODO: Convert images to GrayScale
    # Input images can be either RGB or Grayscale uint8 (0 -> 255). Both need
    # to be supported.
    # Input images must be converted to normalized Grayscale (0.0 -> 1.0)
    # skimage.color.rgb2gray may be useful if the input is RGB.
    if len(I1.shape) == 3 and I1.shape[2] == 3:
        I1 = skimage.color.rgb2gray(I1)
    if len(I2.shape) == 3 and I2.shape[2] == 3:
        I2 = skimage.color.rgb2gray(I2)

    # TODO: Detect features in both images
    locs1 = corner_detection(I1, sigma)
    locs2 = corner_detection(I2, sigma)

    # TODO: Obtain descriptors for the computed feature locations
    desc1, locs1 = computeBrief(I1, locs1)
    desc2, locs2 = computeBrief(I2, locs2)

    # TODO: Match features using the descriptors
    matches = briefMatch(desc1, desc2, ratio)


    # FIX: Extract only the matched keypoints
    # matches[:, 0] contains indices into locs1
    # matches[:, 1] contains indices into locs2
    locs1_matched = locs1[matches[:, 0]]
    locs2_matched = locs2[matches[:, 1]]


    # ==== end of code ====

    return matches, locs1_matched, locs2_matched
```

Design a fix to make BRIEF more rotation invariant. Feel free to make any helper functions as necessary. But you cannot use any additional OpenCV or Scikit-Image functions.

```python
def imageOrientation(image):
    print(image.shape)

    if len(image.shape) == 3:
        # Convert to grayscale if image is RGB
        image = cv2.cvtColor(image, cv2.COLOR_RGB2GRAY)
    
    # Compute gradients
    gy, gx = np.gradient(image.astype(np.float64))  # ensure float for accuracy
    
    # Compute gradient magnitude and direction
    magnitude = np.sqrt(gx**2 + gy**2)
    direction = np.arctan2(gy, gx)  # in radians, range [-π, π]
    
    # Filter out weak gradients (they add noise)
    threshold = np.percentile(magnitude, 50)  # keep top 50% of gradients
    strong_mask = magnitude > threshold

    direction = direction[strong_mask]
    magnitude = magnitude[strong_mask]
    
    # Convert to degrees [0, 360)
    direction_deg = (np.degrees(direction) + 180) % 360
    
    # Bin the directions into discrete angles
    hist, bins = np.histogram(direction_deg, bins=36, range=(0, 360), weights = magnitude)
    
    # Find dominant direction - use CENTER of the bin, not the edge!
    dominant_bin_idx = np.argmax(hist)
    dominant_direction = (bins[dominant_bin_idx] + bins[dominant_bin_idx + 1]) / 2.0
    
    print(f"Dominant direction: {dominant_direction:.1f}°")
    print(f"Histogram max count: {hist[dominant_bin_idx]} out of {np.sum(strong_mask)} strong gradients")
    
    return dominant_direction


def matchPicsRotInvariant(image, rotated_image, ratio, sigma):
    image_dom_grad = imageOrientation(image)
    rotated_dom_grad = imageOrientation(rotated_image)

    # Calculate angle difference
    angle_diff = image_dom_grad - rotated_dom_grad
    
    # Normalize angle to [-180, 180] range
    while angle_diff > 180:
        angle_diff -= 360
    while angle_diff < -180:
        angle_diff += 360
    
    print(f"Rotating image by: {angle_diff:.1f}°")
    
    # Rotate the rotated image back to align with original
    rotated_image = scipy.ndimage.rotate(rotated_image, angle_diff, reshape=False)

    matches, locs1, locs2 = matchPics(image, rotated_image, ratio, sigma)
    return matches, locs1, locs2, image, rotated_image
```

```python
def computeH(x1, x2):
    """
    Compute the homography between two sets of points

    Input
    -----
    x1, x2: Sets of points

    Returns
    -------
    H2to1: 3x3 homography matrix that best transforms x2 to x1
    """

    if x1.shape != x2.shape:
        raise RuntimeError('number of points do not match')

    # ===== your code here! =====
    # TODO: Compute the homography between two sets of points
    if x1.shape != x2.shape:
        raise RuntimeError('number of points do not match')

    N = x1.shape[0]
    A = []
    for i in range(N):
        x1_i, y1_i = x1[i]
        x2_i, y2_i = x2[i]
        A.append([-x2_i, -y2_i, -1, 0, 0, 0, x1_i*x2_i, x1_i*y2_i, x1_i])
        A.append([0, 0, 0, -x2_i, -y2_i, -1, y1_i*x2_i, y1_i*y2_i, y1_i])
        
    A = np.array(A)

    U, D, V = np.linalg.svd(A)
    H2to1 = V[-1].reshape(3, 3)
    
    # ==== end of code ====

    return H2to1
```

### Q2.2.2 (10 points):

Implement the function computeH_norm

```python
def computeH_norm(x1, x2):
    """
    Compute the homography between two sets of points using normalization

    Input
    -----
    x1, x2: Sets of points

    Returns
    -------
    H2to1: 3x3 homography matrix that best transforms x2 to x1
    """

    # ===== your code here! =====

    # TODO: Compute the centroid of the points
    centroid1 = np.mean(x1, axis=0)
    centroid2 = np.mean(x2, axis=0)

    # TODO: Shift the origin of the points to the centroid
    x1_shifted = x1 - centroid1
    x2_shifted = x2 - centroid2

    # TODO: Normalize the points so that the largest distance from the
    # origin is equal to sqrt(2)
    # Max distance using L2 norm
    max_dist1 = np.max(np.sqrt(np.sum(x1_shifted**2, axis=1)))
    max_dist2 = np.max(np.sqrt(np.sum(x2_shifted**2, axis=1)))

    scale1 = np.sqrt(2) / max_dist1
    scale2 = np.sqrt(2) / max_dist2

    x1_normalized = x1_shifted * scale1
    x2_normalized = x2_shifted * scale2

    # TODO: Similarity transform 1
    T1 = np.array([[scale1, 0,      -scale1*centroid1[0]],
                   [0,      scale1, -scale1*centroid1[1]],
                   [0,      0,      1]])

    # TODO: Similarity transform 2
    T2 = np.array([[scale2, 0,      -scale2*centroid2[0]],
                   [0,      scale2, -scale2*centroid2[1]],
                   [0,      0,      1]])

    # TODO: Compute homography
    H = computeH(x1_normalized, x2_normalized)

    # TODO: Denormalization
    H2to1 = np.linalg.inv(T1) @ H @ T2

    # ==== end of code ====

    return H2to1
```

### Q2.2.3 (25 points):

Implement RANSAC

```python
def computeH_ransac(locs1, locs2, max_iters, inlier_tol):
    """
    Estimate the homography between two sets of points using ransac

    Input
    -----
    locs1, locs2: Lists of points
    max_iters: the number of iterations to run RANSAC for
    inlier_tol: the tolerance value for considering a point to be an inlier

    Returns
    -------
    bestH2to1: 3x3 homography matrix that best transforms locs2 to locs1
    inliers: indices of RANSAC inliers

    """

    # ===== your code here! =====
    # Make sure locs1 and locs2 have the same number of points
    assert locs1.shape[0] == locs2.shape[0], f"locs1 and locs2 must have same number of points, got {locs1.shape[0]} and {locs2.shape[0]}"
    
    # TODO:
    # Compute the best fitting homography using RANSAC
    # given a list of matching points locs1 and loc2
    best_inliers_idx = []
    bestH2to1 = None
    max_inlier_count = 0

    for i in range(max_iters):
        # Randomly select 4 points
        sample_indices = np.random.choice(range(len(locs1)), 4, replace=False) # Sample array, output size 4, no replacement (can't have same point twice)
        sample_locs1 = locs1[sample_indices]
        sample_locs2 = locs2[sample_indices]

        # Compute homography using the sample points
        H2to1 = computeH_norm(sample_locs1, sample_locs2)

        # Check how many points are inliers
        # Convert to homogenous coordinates
        locs1_hom = np.hstack([locs1, np.ones((locs1.shape[0], 1))])
        locs2_hom = np.hstack([locs2, np.ones((locs2.shape[0], 1))])
        # Project locs2 onto locs1 using H2to1
        locs1_proj = H2to1 @ locs2_hom.T
        locs1_proj /= locs1_proj[2, :] # Convert to inhomogeneous coordinates
        # Compute squared error between projected points and actual points
        error = np.sum((locs1_proj[:2, :] - locs1.T)**2, axis=0)
        inliers_idx = np.where(error < inlier_tol**2)[0]

        # Update best homography if current one has more inliers
        if len(inliers_idx) > max_inlier_count:
            max_inlier_count = len(inliers_idx)
            bestH2to1 = H2to1
            best_inliers_idx = inliers_idx

    # Convert to binary vector
    best_inliers = np.zeros(len(locs1))
    best_inliers[best_inliers_idx] = 1
            
    # ==== end of code ====

    return bestH2to1, best_inliers
```

Explain the effect of max_iters and inlier_tol:

**max_iters (Maximum Iterations):**
Higher max_iters gives RANSAC more opportunities to sample different point sets and find a better homography. However, there are diminishing returns - after a certain threshold (around 500-1000 iterations), the quality improvement becomes negligible as RANSAC has likely already found the optimal solution. Very low iterations (< 100) may result in poor alignment, while values beyond 1000 provide minimal additional benefit.

**inlier_tol (Inlier Tolerance):**
This parameter requires a balanced value. When inlier_tol is too low (< 0.5 pixels), the threshold is overly strict, rejecting valid matches and resulting in too few inliers, which leads to an unstable homography. When inlier_tol is too high (> 5 pixels), it becomes too permissive and incorrectly accepts outliers as inliers, corrupting the homography and causing distortions. The optimal range is around 1-3 pixels, where genuine correspondences are accepted while mismatches are filtered out.
